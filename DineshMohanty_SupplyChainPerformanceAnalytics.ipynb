## 6. Data Preprocessing

### 6.1 Missing Value Analysis

Missing values can affect calculations and visualizations. Therefore, missing values are identified before deciding whether columns should be removed, retained, or appropriately handled.

```python
missing_values = df.isnull().sum()

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": missing_percentage
})

missing_summary = missing_summary[
    missing_summary["Missing Values"] > 0
].sort_values(
    by="Missing Percentage",
    ascending=False
)

missing_summary
```

```python
df.isnull().sum()
```

### 6.2 Duplicate Record Analysis

Duplicate records are checked to determine whether identical rows exist in the dataset.

```python
duplicate_count = df.duplicated().sum()

print(f"Number of duplicate rows: {duplicate_count}")
```

### 6.3 Order-Level Structure

The dataset contains order-item level records. Therefore, a single order can contain multiple product line items. The number of unique orders is compared with the total number of records to understand the dataset's analytical grain.

```python
total_rows = len(df)
unique_orders = df["Order Id"].nunique()
unique_customers = df["Customer Id"].nunique()
unique_products = df["Product Card Id"].nunique()

print(f"Total records: {total_rows:,}")
print(f"Unique orders: {unique_orders:,}")
print(f"Unique customers: {unique_customers:,}")
print(f"Unique products: {unique_products:,}")
```

```python
average_items_per_order = total_rows / unique_orders

print(f"Average line items per order: {average_items_per_order:.2f}")
```

```python
# Overall delivery performance

delivery_status = df["Delivery Status"].value_counts()

delivery_status
```

```python
delivery_percentage = df["Delivery Status"].value_counts(normalize=True) * 100

delivery_percentage.round(2)
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Delivery Status"
)

plt.title("Delivery Status Distribution")
plt.xlabel("Delivery Status")
plt.ylabel("Number of Records")
plt.xticks(rotation=20)

plt.show()
```

```python
# Which shipping modes have the highest delivery delay rate?

shipping_performance = (
    df.groupby("Shipping Mode")["Late_delivery_risk"]
    .mean()
    .mul(100)
    .reset_index()
)

shipping_performance.columns = [
    "Shipping Mode",
    "Late Delivery Rate (%)"
]

shipping_performance = shipping_performance.sort_values(
    "Late Delivery Rate (%)",
    ascending=False
)

shipping_performance
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=shipping_performance,
    x="Shipping Mode",
    y="Late Delivery Rate (%)"
)

plt.title("Late Delivery Rate by Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Late Delivery Rate (%)")

plt.show()
```

```python
# Regional Delivery Performance

region_delivery = (
    df.groupby("Order Region")["Late_delivery_risk"]
    .mean()
    .mul(100)
    .reset_index()
)

region_delivery.columns = [
    "Order Region",
    "Late Delivery Rate (%)"
]

region_delivery = region_delivery.sort_values(
    "Late Delivery Rate (%)",
    ascending=False
)

region_delivery
```

```python
top_regions = region_delivery.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_regions,
    x="Late Delivery Rate (%)",
    y="Order Region"
)

plt.title("Top 10 Regions by Late Delivery Rate")
plt.xlabel("Late Delivery Rate (%)")
plt.ylabel("Order Region")

plt.show()
```

```python
# Market Performance

market_delivery = (
    df.groupby("Market")["Late_delivery_risk"]
    .mean()
    .mul(100)
    .reset_index()
)

market_delivery.columns = [
    "Market",
    "Late Delivery Rate (%)"
]

market_delivery = market_delivery.sort_values(
    "Late Delivery Rate (%)",
    ascending=False
)

market_delivery
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=market_delivery,
    x="Market",
    y="Late Delivery Rate (%)"
)

plt.title("Late Delivery Rate by Market")
plt.xlabel("Market")
plt.ylabel("Late Delivery Rate (%)")

plt.show()
```

### Business Finding – Overall Delivery Performance

The analysis shows that late delivery is the largest delivery-status category, accounting for 54.83% of the records. Advance shipping accounts for 23.04%, while 17.84% of records represent shipping on time. The remaining 4.30% are classified as shipping canceled.

This indicates that delivery delays are an important area of supply chain performance that requires further analysis. The following analysis will examine shipping modes, regions, markets, and product categories to identify where delivery problems are more concentrated.

#  Key Findings

The major findings from the supply chain analysis are:

### Delivery Performance
- Late delivery was the largest delivery-status category, accounting for approximately 54.83% of the records.
- Delivery performance varies across different shipping modes, regions, markets, and product categories.
- Some geographical areas and shipping modes show higher late-delivery rates than others.

### Product & Profitability
- Product categories differ in their sales and profitability performance.
- A small group of products contributes significantly to overall sales and profit.
- High sales does not necessarily mean the same product or category has the highest profitability.

### Discount Analysis
- Discount levels show a measurable relationship with sales and profitability.
- The relationship should be interpreted as an association rather than a direct cause-and-effect relationship.

### Customer & Geographic Performance
- Sales and profitability vary across different regions and customer segments.
- Certain regions and customer segments contribute a larger share of overall business performance.

### Time-Based Performance
- Monthly sales, profit, and late-delivery rates show changes over the analyzed period.
- Comparing these trends helps identify periods where business performance and delivery performance changed.

### Overall Supply Chain Observation
- Delivery performance is an important area requiring management attention.
- Shipping modes, geographical regions, product categories, and time periods can be further examined to identify opportunities for improving supply chain efficiency.
